## The real data

The readings above are made up. This cell loads a real dataset: every injury
road collision the police recorded in West Yorkshire over the years the
module uses. `describe` tells you how many rows there are, which years they
actually cover, and the credit you must include if you publish anything made
from them. Read the years it prints rather than trusting a sentence that
claims a span: the data decides.


```python
from cwai import registry

collisions = data.load_collisions()
assert not collisions.empty, "The collision data loaded but contains no rows."
# The module's years and the file's years must agree; if they do not, the
# data folder is stale or was built from a different set of downloads.
assert data.available_years("collisions") == list(registry.YEARS), (
    "The collision data does not cover the years this module expects."
)
print(data.describe("collisions"))
collisions[["accident_year", "date", "severity_label", "number_of_casualties"]].head()
```
