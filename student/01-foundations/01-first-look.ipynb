## A mistake that looked like it worked

Severity came to you as words. The casualty type did not. The column
`casualty_type` holds a number code for each person hurt. One code means a
person walking, another means a person cycling, and so on.

The next cell comes from a real session with this data. It was written to
count the injured cyclists. It ran, and it gave a number. Read it before you
run it.

```python
# The first three casualty codes, in the order the STATS19 guide lists them.
casualty_names = ["Pedestrian", "Cyclist", "Motorcycle rider"]

# Pair each name with its code. enumerate numbers the names as it goes.
names_by_code = dict(enumerate(casualty_names, start=1))

casualties["casualty_name"] = casualties["casualty_type"].map(names_by_code)
cyclists = casualties[casualties["casualty_name"] == "Cyclist"]
len(cyclists)
```

The answer came out as 120.

West Yorkshire has about 2.3 million people, and this table covers five years.
Over that time, 120 injured cyclists is far too few. The number was wrong, and
it looked wrong. That is the only reason anybody checked it.

## The fix, and the lesson in it

Python counts positions from 0, not from 1. The first name in the list sits at
position 0. `enumerate` numbers from 0 unless you tell it otherwise, and
`start=1` told it otherwise.

So every code was shifted by one. People cycling, code 1, were labelled
`"Pedestrian"`. The count of cyclists then picked up code 2, which is a
different group of people.

In this data `casualty_type` is 0 for a person walking and 1 for a person
cycling. Here is the same code with `start=1` taken out.

```python
names_by_code = dict(enumerate(casualty_names))

casualties["casualty_name"] = casualties["casualty_type"].map(names_by_code)
cyclists = casualties[casualties["casualty_name"] == "Cyclist"]
len(cyclists)
```

2,304 cyclists, not 120. Nothing in the first version looked broken. It read
well, it ran, and it gave a number. Checking a number against what you already
know is the only guard you have.
